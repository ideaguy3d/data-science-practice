```python
# leet3844
class Solution:
    def almostPalindromic(self, s: str) -> int:
        n = len(s)
        best = 0

        def expand(l, r):
            used = False
            start_l, start_r = l, r

            while l >= 0 and r < n:
                if s[l] == s[r]:
                    l -= 1
                    r += 1
                elif not used:
                    # try skipping left OR right once
                    skip_left = expand_one(l - 1, r)
                    skip_right = expand_one(l, r + 1)
                    return max(skip_left, skip_right)
                else:
                    break

            return r - l - 1

        def expand_one(l, r):
            while l >= 0 and r < n and s[l] == s[r]:
                l -= 1
                r += 1
            return r - l - 1

        for i in range(n):
            # odd length center
            best = max(best, expand(i, i))
            # even length center
            best = max(best, expand(i, i + 1))

        return best


c = Solution()
# 1st test case
assert c.almostPalindromic("abcba") == 5
# 2nd test case
assert c.almostPalindromic("abcd") == 2
# 3rd test case
assert c.almostPalindromic("abccba") == 6
# 4th test case
assert c.almostPalindromic("zzabba") == 5
# 5th test case
assert c.almostPalindromic("aabbc") == 3
```
